# Local Jobs Agent: API smoke test

This notebook checks that your OpenAI API key can make a small model request, then demonstrates the local job-search tool. The job records are mock examples, not live vacancies.

Before running the cells, open a terminal in `aiLocaljobs` and run `uv sync`. Copy `.env.example` to `.env` and put your key in `.env`. **The notebook checks that a key is present but never displays it.** Select the project's `.venv` Python kernel if VS Code asks.

```python
from pathlib import Path
import os
import sys

from dotenv import load_dotenv
from openai import OpenAI


def find_project_root(start: Path) -> Path:
    for folder in (start, *start.parents):
        if (folder / "pyproject.toml").is_file() and (folder / "src" / "ai_local_jobs").is_dir():
            return folder
        nested_project = folder / "aiLocaljobs"
        if (nested_project / "pyproject.toml").is_file() and (nested_project / "src" / "ai_local_jobs").is_dir():
            return nested_project
    raise FileNotFoundError("Could not find the aiLocaljobs project folder.")


project_root = find_project_root(Path.cwd().resolve())
load_dotenv(project_root / ".env")

if not os.getenv("OPENROUTER_API_KEY"):
    raise RuntimeError("OPENROUTER_API_KEY is missing. Add it to aiLocaljobs/.env and rerun this cell.")

sys.path.insert(0, str(project_root / "src"))
from ai_local_jobs.agent import create_job_agent

agent = create_job_agent()
result = agent.invoke(
    {"messages": [{"role": "user", "content": "Reply exactly: API connection works."}]}
)
print(result["messages"][-1].content)
```
